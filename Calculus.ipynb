## Importing Libraries

```python
import math
import cmath
```

The following helper functions compute numerical approximations using the central difference method. They are reused throughout this notebook.

The `deriv` function approximates the derivative of a single-variable function at a point using the central difference formula:

$$ f'(x) \approx \frac{f(x+h) - f(x-h)}{2h} $$

```python
def deriv(f, x, h=1e-7):
    return (f(x + h) - f(x - h)) / (2 * h)

# verify: derivative of sin(x) at x = 0 is cos(0) = 1
deriv(math.sin, 0)
```

The `partial_deriv` function approximates the partial derivative of a multivariable function with respect to its $i$-th argument at a given point, using the central difference formula:

$$ \frac{\partial f}{\partial x_i} \approx \frac{f(x_1, \ldots, x_i+h, \ldots, x_n) - f(x_1, \ldots, x_i-h, \ldots, x_n)}{2h} $$

Parameters: `f` (callable), `point` (sequence of float), `i` (zero-based index of variable), `h` (step size, default `1e-7`).

```python
def partial_deriv(f, point, i, h=1e-7):
    x_fwd = list(point)
    x_bwd = list(point)
    x_fwd[i] += h
    x_bwd[i] -= h
    return (f(*x_fwd) - f(*x_bwd)) / (2 * h)

# verify: d/dx(x * y) at (2, 3) = y = 3
partial_deriv(lambda x, y: x * y, [2, 3], 0)
```

<u>**Limits**</u>

The principal notation for limits and the behavior of functions as variables approach specific values or grow without bound.

<a id='calculuslimitslimit'></a>

---

<u>**Limit of a Function**</u>

The limit of a function $f(x)$ as $x$ approaches a value $a$ is the value that $f(x)$ gets arbitrarily close to as $x$ gets arbitrarily close to $a$. It is denoted by:

$$\lim_{x \to a} f(x)$$

If this limit equals some value $L$, the statement is written:

$$\lim_{x \to a} f(x) = L$$

```python
# lim_{x -> 0} sin(x) / x = 1
f = lambda x: math.sin(x) / x
for x in [1.0, 0.1, 0.01, 0.001, 0.0001]:
    print(f'x = {x:<8}  f(x) = {f(x):.10f}')
```

<a id='calculuslimitsonesided'></a>

---

<u>**One-Sided Limits**</u>

A one-sided limit considers the behavior of $f(x)$ as $x$ approaches $a$ from only one direction. The left-hand limit is denoted:

$$\lim_{x \to a^{-}} f(x) = \lim_{x \uparrow a} f(x)$$

The right-hand limit is denoted:

$$\lim_{x \to a^{+}} f(x) = \lim_{x \downarrow a} f(x)$$

The two-sided limit $\lim_{x \to a} f(x)$ exists if and only if both one-sided limits exist and are equal.

```python
# f(x) = x / |x| has different one-sided limits at x = 0
f = lambda x: x / abs(x)
print('Approaching 0 from the right (+):')
for x in [1.0, 0.1, 0.01, 0.001]:
    print(f'  x = {x:<8}  f(x) = {f(x):+.1f}')
print('Approaching 0 from the left (-):')
for x in [-1.0, -0.1, -0.01, -0.001]:
    print(f'  x = {x:<8}  f(x) = {f(x):+.1f}')
```

<a id='calculuslimitsinfinity'></a>

---

<u>**Limits at Infinity**</u>

The limit of $f(x)$ as $x$ increases or decreases without bound is written:

$$\lim_{x \to \infty} f(x) \qquad \text{and} \qquad \lim_{x \to -\infty} f(x)$$

```python
# lim_{x -> inf} (1 + 1/x)^x = e
f = lambda x: (1 + 1 / x) ** x
for x in [10, 100, 1000, 10000, 100000]:
    print(f'x = {x:<8}  f(x) = {f(x):.10f}')
print(f'{"e":>19} = {math.e:.10f}')
```

<a id='calculuslimitssequence'></a>

---

<u>**Limit of a Sequence**</u>

For a sequence $a_1, a_2, a_3, \ldots$ the notation:

$$\lim_{n \to \infty} a_n$$

expresses the limiting value of the sequence. If the limit equals $L$, the sequence is said to converge to $L$.

```python
# a_n = 1/n converges to 0
a = lambda n: 1 / n
for n in [1, 10, 100, 1000, 10000, 100000]:
    print(f'n = {n:<8}  a_n = {a(n):.10f}')
```

<a id='calculuslimitslimsup'></a>

---

<u>**Limit Superior and Limit Inferior**</u>

For a sequence $a_1, a_2, a_3, \ldots$ the limit superior and limit inferior are defined as:

$$\limsup_{n \to \infty} a_n = \lim_{n \to \infty} \left[ \sup\{a_k : k \ge n\} \right]$$

$$\liminf_{n \to \infty} a_n = \lim_{n \to \infty} \left[ \inf\{a_k : k \ge n\} \right]$$

These may also be written as $\overline{\lim}$ for $\limsup$ and $\underline{\lim}$ for $\liminf$. If the sequence converges, then $\limsup = \liminf = \lim$.

```python
# a_n = (-1)^n * (1 + 1/n) oscillates: lim sup = 1, lim inf = -1
a = [(-1)**n * (1 + 1 / n) for n in range(1, 20001)]
for n in [10, 100, 1000, 10000]:
    tail = a[n - 1:]
    print(f'n = {n:<6}  sup = {max(tail):+.6f}   inf = {min(tail):+.6f}')
```

<a id='calculusderivatives'></a>

---

<u>**Derivatives (Single Variable)**</u>

Notation for derivatives of functions of a single independent variable. Covers scalar-valued and vector-valued functions.

<a id='calculusderivativesnotation'></a>

---

<u>**Derivative Notation**</u>

Given a function $f$, two standard notations exist for the derivative of $f$.

Newton's notation uses a prime mark:

$$f'$$

Leibniz's notation uses the differential operator:

$$\frac{df}{dx}$$

Some authors use a roman (upright) d in Leibniz's notation, as in $\frac{\mathrm{d}f}{\mathrm{d}x}$, to distinguish the differential operator from the variable $d$. Both notations express the rate of change of $f$ with respect to $x$.

```python
# f(x) = x^2, so f'(x) = df/dx = 2x
f = lambda x: x ** 2
x = 3.0
print(f"f({x})  = {f(x)}")
print(f"f'({x}) = {deriv(f, x):.6f}")
print(f"exact   = {2 * x}")
```

<a id='calculusderivativeshigherorder'></a>

---

<u>**Higher Order Derivatives**</u>

Higher order derivatives in Newton's notation are expressed with additional prime marks. The second derivative is $f''$, the third is $f'''$, and so forth. Sometimes lower case roman numerals replace multiple prime marks, e.g. $f^{(\mathrm{iv})}$. For a positive integer $n$, the $n$-th derivative may be written $f^{(n)}$.

In Leibniz's notation, higher order derivatives are expressed with exponents:

$$\frac{df}{dx}, \quad \frac{d^2 f}{dx^2}, \quad \frac{d^3 f}{dx^3}, \quad \ldots$$

The $n$-th derivative is written $\frac{d^n f}{dx^n}$.

```python
# f(x) = x^4: f' = 4x^3, f'' = 12x^2, f''' = 24x
f = lambda x: x ** 4
x, h = 2.0, 1e-3
f_1 = (f(x + h) - f(x - h)) / (2 * h)
f_2 = (f(x + h) - 2*f(x) + f(x - h)) / h**2
f_3 = (f(x+2*h) - 2*f(x+h) + 2*f(x-h) - f(x-2*h)) / (2*h**3)
print(f"f'({x})   = {f_1:.4f}   (exact: {4*x**3})")
print(f"f''({x})  = {f_2:.4f}   (exact: {12*x**2})")
print(f"f'''({x}) = {f_3:.4f}   (exact: {24*x})")
```

<a id='calculusderivativesevaluation'></a>

---

<u>**Evaluation at a Point**</u>

The value of a derivative at a specific point $x = a$ is written as $f'(a)$ in Newton's notation. In Leibniz's notation, the evaluation bar is used:

$$\left. \frac{df}{dx} \right|_{x=a}$$

If $f(x) = e^x$, then $f'(x) = e^x$ and:

$$\left. \frac{df}{dx} \right|_{x=0} = e^0 = 1$$

```python
# f(x) = e^x, f'(a) = e^a
f = lambda x: math.exp(x)
for a in [0, 1, 2]:
    print(f"df/dx|_(x={a}) = {deriv(f, a):.10f}   (exact: {math.exp(a):.10f})")
```

<a id='calculusderivativesdot'></a>

---

<u>**Dot Notation**</u>

When the independent variable is time $t$, a dot above the function denotes the derivative. If $y$ is a function of $t$, then $\dot{y}$ denotes $dy/dt$. Likewise $\ddot{y}$ is the second derivative $d^2 y / dt^2$:

$$\dot{y} = \frac{dy}{dt}, \qquad \ddot{y} = \frac{d^2 y}{dt^2}$$

This notation is standard in physics and engineering.

```python
# position y(t) = 5t^2, velocity y_dot = 10t, acceleration y_ddot = 10
y = lambda t: 5 * t ** 2
t, h = 3.0, 1e-4
y_dot = (y(t + h) - y(t - h)) / (2 * h)
y_ddot = (y(t + h) - 2*y(t) + y(t - h)) / h**2
print(f'y({t})     = {y(t)}')
print(f'y_dot({t}) = {y_dot:.6f}   (exact: {10 * t})')
print(f'y_ddot     = {y_ddot:.4f}   (exact: 10)')
```

<a id='calculusderivativesoperator'></a>

---

<u>**D Operator Notation**</u>

The notation $\frac{d}{dx}$ can be viewed as an operator applied to a function. It is also denoted with a capital $D$:

$$\frac{d}{dx}\left[ x^2 - 3x + 2 \right] = D\left[ x^2 - 3x + 2 \right] = 2x - 3$$

Higher order derivatives are denoted $\frac{d^n}{dx^n}$ or $D^n$.

```python
# D[x^2 - 3x + 2] = 2x - 3
f = lambda x: x**2 - 3*x + 2
x = 5.0
print(f'D[x^2 - 3x + 2] at x = {x}: {deriv(f, x):.6f}   (exact: {2*x - 3})')
```

<a id='calculusderivativesvectorvalued'></a>

---

<u>**Vector-Valued Derivative**</u>

The derivative of a vector-valued function of a single variable is defined component-wise. For $f : \mathbb{R} \to \mathbb{R}^n$, the derivative is denoted either $f'$ or $df/dt$. Higher order derivatives are $f''$ and $d^2 f/dt^2$.

$$f(t) = \begin{bmatrix} f_1(t) \\ f_2(t) \\ \vdots \\ f_n(t) \end{bmatrix}, \qquad f'(t) = \begin{bmatrix} f_1'(t) \\ f_2'(t) \\ \vdots \\ f_n'(t) \end{bmatrix}$$

```python
# f(t) = (cos t, sin t), f'(t) = (-sin t, cos t)
f1 = lambda t: math.cos(t)
f2 = lambda t: math.sin(t)
t = math.pi / 4
f_prime = [deriv(f1, t), deriv(f2, t)]
exact = [-math.sin(t), math.cos(t)]
print(f"f'(pi/4) = [{f_prime[0]:.8f}, {f_prime[1]:.8f}]")
print(f"exact    = [{exact[0]:.8f}, {exact[1]:.8f}]")
```

<a id='calculuspartial'></a>

---

<u>**Partial Derivatives (Scalar-Valued)**</u>

When a function depends on two or more variables, the notion of partial derivative comes into play. This section covers partial derivatives of scalar-valued functions $f : \mathbb{R}^n \to \mathbb{R}$.

<a id='calculuspartialpartial'></a>

---

<u>**Partial Derivatives**</u>

For a function $f : \mathbb{R}^n \to \mathbb{R}$, the partial derivative of $f$ with respect to its $i$-th argument is denoted:

$$\frac{\partial f}{\partial x_i}$$

For a function of two variables $x$ and $y$, the partial derivatives can also be written as $f_x = \partial_x f = \partial f / \partial x$ and $f_y = \partial_y f = \partial f / \partial y$. Another notation is $\partial_i f$.

The notation $D_x$ means to take the derivative in the $x$-direction:

$$D_x f = \partial_x f = f_x = \frac{\partial f}{\partial x}$$

```python
# f(x, y) = x^2 * y + y^3
# df/dx = 2xy,  df/dy = x^2 + 3y^2
f = lambda x, y: x**2 * y + y**3
x, y = 2.0, 3.0
df_dx = partial_deriv(f, [x, y], 0)
df_dy = partial_deriv(f, [x, y], 1)
print(f'df/dx at ({x},{y}) = {df_dx:.6f}   (exact: {2*x*y})')
print(f'df/dy at ({x},{y}) = {df_dy:.6f}   (exact: {x**2 + 3*y**2})')
```

<a id='calculuspartialdirectional'></a>

---

<u>**Directional Derivative**</u>

If $\mathbf{v}$ is a unit vector, then $D_\mathbf{v} f$ is the derivative of $f$ in the $\mathbf{v}$ direction. This is the directional derivative, computed as the dot product of the gradient with the direction vector:

$$D_\mathbf{v} f = \nabla f \cdot \mathbf{v}$$

(See: [Gradient](#calculuspartialgradient))

```python
# f(x, y) = x^2 + y^2,  grad f = (2x, 2y)
# directional derivative along v = (1/sqrt(2), 1/sqrt(2))
f = lambda x, y: x**2 + y**2
x, y = 1.0, 1.0
grad = [partial_deriv(f, [x, y], 0), partial_deriv(f, [x, y], 1)]
v = [1 / math.sqrt(2), 1 / math.sqrt(2)]
D_v = sum(g * vi for g, vi in zip(grad, v))
print(f'grad f = [{grad[0]:.4f}, {grad[1]:.4f}]')
print(f'D_v f  = {D_v:.6f}   (exact: {2 * math.sqrt(2):.6f})')
```

<a id='calculuspartialhigherorder'></a>

---

<u>**Higher Order Partial Derivatives**</u>

Higher order partial derivatives are denoted:

$$\frac{\partial^2 f}{\partial x_i \, \partial x_j}$$

which means to first take the partial derivative of $f$ with respect to $x_j$ and then the partial derivative of that result with respect to $x_i$:

$$\frac{\partial^2 f}{\partial x_i \, \partial x_j} = \frac{\partial}{\partial x_i} \left( \frac{\partial f}{\partial x_j} \right)$$

For a function of two variables $x$ and $y$, the four second-order partials are written $f_{xx}$, $f_{xy}$, $f_{yx}$, $f_{yy}$. For most functions encountered in practice, the order of differentiation does not matter ($f_{xy} = f_{yx}$).

```python
# f(x, y) = x^2 * y^3
# f_xx = 2y^3,  f_yy = 6x^2*y,  f_xy = 6xy^2
f = lambda x, y: x**2 * y**3
x, y = 2.0, 3.0
h = 1e-4
f_xx = (f(x + h, y) - 2 * f(x, y) + f(x - h, y)) / h**2
f_yy = (f(x, y + h) - 2 * f(x, y) + f(x, y - h)) / h**2
f_xy = (f(x+h, y+h) - f(x+h, y-h) - f(x-h, y+h) + f(x-h, y-h)) / (4 * h**2)
print(f'f_xx = {f_xx:.4f}  (exact: {2 * y**3})')
print(f'f_yy = {f_yy:.4f}  (exact: {6 * x**2 * y})')
print(f'f_xy = {f_xy:.4f}  (exact: {6 * x * y**2})')
```

<a id='calculuspartialgradient'></a>

---

<u>**Gradient**</u>

The gradient of a function $f : \mathbb{R}^n \to \mathbb{R}$ is the vector of partial derivatives. It is denoted $\operatorname{grad} f$ or $\nabla f$:

$$\nabla f = \begin{bmatrix} \dfrac{\partial f}{\partial x_1} \\[6pt] \dfrac{\partial f}{\partial x_2} \\[6pt] \vdots \\[6pt] \dfrac{\partial f}{\partial x_n} \end{bmatrix}$$

For a function of three variables, the gradient can be expressed in $\mathbf{i}, \mathbf{j}, \mathbf{k}$ notation:

$$\operatorname{grad} f = \nabla f = f_x \, \mathbf{i} + f_y \, \mathbf{j} + f_z \, \mathbf{k}$$

```python
# f(x, y, z) = x^2 + 2yz + z^3
# grad f = (2x, 2z, 2y + 3z^2)
f = lambda x, y, z: x**2 + 2*y*z + z**3
p = [1.0, 2.0, 3.0]
grad = [partial_deriv(f, p, i) for i in range(3)]
exact = [2*p[0], 2*p[2], 2*p[1] + 3*p[2]**2]
print(f'grad f at {p}:')
print(f'  numerical = [{grad[0]:.4f}, {grad[1]:.4f}, {grad[2]:.4f}]')
print(f'  exact     = {exact}')
```

<a id='calculuspartiallaplacian'></a>

---

<u>**Laplacian**</u>

The Laplacian operator is denoted with a capital delta $\Delta$. For $f : \mathbb{R}^n \to \mathbb{R}$:

$$\Delta f = \sum_{i=1}^{n} \frac{\partial^2 f}{\partial x_i^2}$$

For a function of three variables:

$$\Delta f = \frac{\partial^2 f}{\partial x^2} + \frac{\partial^2 f}{\partial y^2} + \frac{\partial^2 f}{\partial z^2}$$

The Laplacian can also be written as $\nabla \cdot (\nabla f) = \nabla^2 f$.

```python
# f(x, y, z) = x^2 + y^2 + z^2
# f_xx = 2, f_yy = 2, f_zz = 2, so Laplacian = 6
f = lambda x, y, z: x**2 + y**2 + z**2
p = [1.0, 2.0, 3.0]
h = 1e-4
laplacian = 0
for i in range(3):
    p_fwd, p_bwd = list(p), list(p)
    p_fwd[i] += h
    p_bwd[i] -= h
    laplacian += (f(*p_fwd) - 2 * f(*p) + f(*p_bwd)) / h**2
print(f'Laplacian at {p} = {laplacian:.6f}   (exact: 6)')
```

<a id='calculuspartialhessian'></a>

---

<u>**Hessian**</u>

For a function $f : \mathbb{R}^n \to \mathbb{R}$, the Hessian matrix $Hf$ is the $n \times n$ matrix of all partial second derivatives:

$$Hf = \begin{bmatrix} \dfrac{\partial^2 f}{\partial x_1^2} & \dfrac{\partial^2 f}{\partial x_1 \, \partial x_2} & \cdots & \dfrac{\partial^2 f}{\partial x_1 \, \partial x_n} \\[6pt] \dfrac{\partial^2 f}{\partial x_2 \, \partial x_1} & \dfrac{\partial^2 f}{\partial x_2^2} & \cdots & \dfrac{\partial^2 f}{\partial x_2 \, \partial x_n} \\[6pt] \vdots & \vdots & \ddots & \vdots \\[6pt] \dfrac{\partial^2 f}{\partial x_n \, \partial x_1} & \dfrac{\partial^2 f}{\partial x_n \, \partial x_2} & \cdots & \dfrac{\partial^2 f}{\partial x_n^2} \end{bmatrix}$$

```python
# f(x, y) = x^3 + x*y^2
# H = [[6x, 2y], [2y, 2x]]
f = lambda x, y: x**3 + x * y**2
x, y = 2.0, 3.0
h = 1e-4
f_xx = (f(x+h, y) - 2*f(x, y) + f(x-h, y)) / h**2
f_yy = (f(x, y+h) - 2*f(x, y) + f(x, y-h)) / h**2
f_xy = (f(x+h,y+h) - f(x+h,y-h) - f(x-h,y+h) + f(x-h,y-h)) / (4*h**2)
print(f'Hessian at ({x}, {y}):')
print(f'  [[{f_xx:.2f}, {f_xy:.2f}],')
print(f'   [{f_xy:.2f}, {f_yy:.2f}]]')
print(f'exact: [[{6*x}, {2*y}], [{2*y}, {2*x}]]')
```

<a id='calculusvectorvalued'></a>

---

<u>**Derivatives (Vector-Valued)**</u>

Derivatives of functions $\mathbf{f} : \mathbb{R}^m \to \mathbb{R}^n$ that map vectors to vectors. The key objects are the Jacobian matrix, curl, and divergence.

<a id='calculusvectorvaluedjacobian'></a>

---

<u>**Jacobian**</u>

The matrix of all partial derivatives of a vector-valued function $\mathbf{f} : \mathbb{R}^m \to \mathbb{R}^n$ is called the Jacobian matrix, denoted $D\mathbf{f}$:

$$D\mathbf{f} = \begin{bmatrix} \dfrac{\partial f_1}{\partial x_1} & \dfrac{\partial f_1}{\partial x_2} & \cdots & \dfrac{\partial f_1}{\partial x_m} \\[6pt] \dfrac{\partial f_2}{\partial x_1} & \dfrac{\partial f_2}{\partial x_2} & \cdots & \dfrac{\partial f_2}{\partial x_m} \\[6pt] \vdots & \vdots & \ddots & \vdots \\[6pt] \dfrac{\partial f_n}{\partial x_1} & \dfrac{\partial f_n}{\partial x_2} & \cdots & \dfrac{\partial f_n}{\partial x_m} \end{bmatrix}$$

If $\mathbf{f} : \mathbb{R}^n \to \mathbb{R}^n$, then $D\mathbf{f}$ is a square matrix and its determinant is also called the Jacobian of $\mathbf{f}$. For a change of coordinates $\mathbf{f}(x, y, z) = (u, v, w)$, the Jacobian determinant is written:

$$\left| \frac{\partial(u, v, w)}{\partial(x, y, z)} \right|$$

```python
# f: R^2 -> R^2,  f(x, y) = (x*y, x^2 + y)
# Jacobian: [[y, x], [2x, 1]]
f1 = lambda x, y: x * y
f2 = lambda x, y: x**2 + y
x, y = 3.0, 2.0

J = []
for fi in [f1, f2]:
    row = []
    for j in range(2):
        row.append(partial_deriv(fi, [x, y], j))
    J.append(row)

print('Jacobian at (3, 2):')
for row in J:
    print(f'  [{row[0]:8.4f}, {row[1]:8.4f}]')
det_J = J[0][0] * J[1][1] - J[0][1] * J[1][0]
print(f'det(J) = {det_J:.4f}   (exact: {y - 2*x**2})')
```

<a id='calculusvectorvaluedcurl'></a>

---

<u>**Curl**</u>

For a function $\mathbf{f} : \mathbb{R}^3 \to \mathbb{R}^3$, the curl is denoted $\operatorname{curl} \mathbf{f}$ or $\nabla \times \mathbf{f}$:

$$\operatorname{curl} \mathbf{f} = \nabla \times \mathbf{f} = \begin{bmatrix} f_{3,2} - f_{2,3} \\ f_{1,3} - f_{3,1} \\ f_{2,1} - f_{1,2} \end{bmatrix}$$

where $f_{i,j} = \frac{\partial f_i}{\partial x_j}$. The notation $\nabla \times \mathbf{f}$ is inspired by the cross product, treating $\nabla$ as the operator vector:

$$\nabla \times \mathbf{f} = \begin{bmatrix} \partial / \partial x_1 \\ \partial / \partial x_2 \\ \partial / \partial x_3 \end{bmatrix} \times \begin{bmatrix} f_1 \\ f_2 \\ f_3 \end{bmatrix}$$

```python
# f(x, y, z) = (y*z, x*z, x*y)
# curl f = (x - x, y - y, z - z) = (0, 0, 0)
f1 = lambda x, y, z: y * z
f2 = lambda x, y, z: x * z
f3 = lambda x, y, z: x * y
p = [1.0, 2.0, 3.0]
curl = [
    partial_deriv(f3, p, 1) - partial_deriv(f2, p, 2),
    partial_deriv(f1, p, 2) - partial_deriv(f3, p, 0),
    partial_deriv(f2, p, 0) - partial_deriv(f1, p, 1)
]
print(f'curl f at {p} = [{curl[0]:.6f}, {curl[1]:.6f}, {curl[2]:.6f}]')
```

<a id='calculusvectorvalueddivergence'></a>

---

<u>**Divergence**</u>

For a function $\mathbf{f} : \mathbb{R}^n \to \mathbb{R}^n$, the divergence is denoted $\operatorname{div} \mathbf{f}$ or $\nabla \cdot \mathbf{f}$:

$$\operatorname{div} \mathbf{f} = \nabla \cdot \mathbf{f} = \sum_{j=1}^{n} \frac{\partial f_j}{\partial x_j}$$

The notation is inspired by the dot product, treating $\nabla$ as the operator vector:

$$\nabla \cdot \mathbf{f} = \begin{bmatrix} \partial / \partial x_1 \\ \partial / \partial x_2 \\ \partial / \partial x_3 \end{bmatrix} \cdot \begin{bmatrix} f_1 \\ f_2 \\ f_3 \end{bmatrix}$$

For a scalar-valued function $f$, the Laplacian (see: [Laplacian](#calculuspartiallaplacian)) can be expressed as $\nabla \cdot (\nabla f) = \nabla^2 f = \Delta f$.

```python
# f(x, y, z) = (x^2, y^2, z^2),  div f = 2x + 2y + 2z
f1 = lambda x, y, z: x**2
f2 = lambda x, y, z: y**2
f3 = lambda x, y, z: z**2
p = [1.0, 2.0, 3.0]
div = sum(partial_deriv(fi, p, i) for i, fi in enumerate([f1, f2, f3]))
print(f'div f at {p} = {div:.6f}   (exact: {2 * sum(p)})')
```

<a id='calculusintegration'></a>

---

<u>**Integration**</u>

Notation for indefinite and definite integrals, multiple integrals, integration over domains, and line and contour integrals.

<a id='calculusintegrationindefinite'></a>

---

<u>**Indefinite Integral**</u>

Let $f : \mathbb{R} \to \mathbb{R}$. The notation $\int f(x)\, dx$ is the indefinite integral of $f$; it is a function $F$ whose derivative is $f$. The function $F$ is called an antiderivative or a primitive of $f$:

$$\int f(x)\, dx = F(x) + C \qquad \text{where} \quad F'(x) = f(x)$$

Some authors use a roman d in the integral, as in $\int f(x)\, \mathrm{d}x$.

```python
# F(x) = x^3 / 3 is the antiderivative of f(x) = x^2
# verify: F'(x) = x^2
F = lambda x: x**3 / 3
f = lambda x: x**2
x = 5.0
print(f'F(x) = x^3/3,  F({x}) = {F(x):.4f}')
print(f"F'({x}) = {deriv(F, x):.6f}")
print(f'f({x})  = {f(x)}')
```

<a id='calculusintegrationdefinite'></a>

---

<u>**Definite Integral**</u>

The definite integral of $f$ from $a$ to $b$ is denoted:

$$\int_a^b f(x)\, dx$$

Its value is given by the Fundamental Theorem of Calculus:

$$\int_a^b f(x)\, dx = F(x) \Big|_a^b = F(b) - F(a)$$

where $F$ is any antiderivative of $f$. The left and right endpoints may be $-\infty$ and $\infty$, respectively.

```python
# numerical integration via midpoint Riemann sum
def integrate(f, a, b, n=100000):
    dx = (b - a) / n
    return sum(f(a + (i + 0.5) * dx) for i in range(n)) * dx

# integral of x^2 from 0 to 3 = [x^3/3]_0^3 = 9
result = integrate(lambda x: x**2, 0, 3)
print(f'integral of x^2 from 0 to 3 = {result:.8f}   (exact: 9.0)')
```

<a id='calculusintegrationmultiple'></a>

---

<u>**Multiple Integrals**</u>

Multiple integrals extend integration to functions of several variables. A double integral is written:

$$\iint f(x, y)\, dx\, dy$$

This means to integrate $f(x, y)$ first with respect to $x$ (holding $y$ constant), then integrate the result with respect to $y$. It is equivalent to the iterated integral:

$$\int \left[ \int f(x, y)\, dx \right] dy$$

```python
# double integral of x*y over [0,1] x [0,1] = 1/4
n = 300
dx = dy = 1.0 / n
total = 0
for i in range(n):
    x = (i + 0.5) * dx
    for j in range(n):
        y = (j + 0.5) * dy
        total += x * y * dx * dy
print(f'integral of x*y over [0,1]^2 = {total:.8f}   (exact: {1/4})')
```

<a id='calculusintegrationdomain'></a>

---

<u>**Domain Subscripts**</u>

Subscripts on integrals indicate the domain of integration. The typical notation is $\int_A f\, ds$ where $A$ is a set. Some examples:

$$\int_{[0,1]} x^2\, dx \quad \text{means} \quad \int_0^1 x^2\, dx$$

$$\int_{\mathbb{R}} \exp\{-x^2\}\, dx \quad \text{means} \quad \int_{-\infty}^{\infty} e^{-x^2}\, dx$$

$$\int_{[0,1]^2} (x - y)^2\, dx\, dy \quad \text{means} \quad \int_0^1 \int_0^1 (x - y)^2\, dx\, dy$$

```python
# integral of exp(-x^2) over R = sqrt(pi)
# approximate by integrating over [-10, 10]
f = lambda x: math.exp(-x**2)
result = integrate(f, -10, 10)
print(f'integral of exp(-x^2) over R = {result:.8f}')
print(f'sqrt(pi)                     = {math.sqrt(math.pi):.8f}')
```

<a id='calculusintegrationline'></a>

---

<u>**Line Integral and Contour Integral**</u>

When the subscript on the integral is a curve $\gamma$:

$$\int_\gamma f\, ds$$

this denotes the line integral of $f$ along the curve $\gamma$. If the curve is closed (it begins and ends at the same point), the contour integral notation is used:

$$\oint_\gamma f\, ds$$

When the contour is the boundary of a domain $D$, the notation may be:

$$\int_{\partial D} f\, ds \qquad \text{or} \qquad \oint_{\partial D} f\, ds$$

where $\partial D$ indicates the boundary of $D$.

```python
# line integral of f(x,y) = x^2 + y^2 along the unit circle
# parametrize: x = cos(t), y = sin(t), ds = dt
# integral = integral_0^{2pi} (cos^2 t + sin^2 t) dt = 2*pi
n = 100000
dt = 2 * math.pi / n
total = sum(
    (math.cos((i+0.5)*dt)**2 + math.sin((i+0.5)*dt)**2) * dt
    for i in range(n)
)
print(f'line integral of x^2+y^2 along unit circle = {total:.8f}')
print(f'exact (2*pi) = {2 * math.pi:.8f}')
```

<a id='calculustransforms'></a>

---

<u>**Convolution and Transforms**</u>

Convolution combines two functions into a new function. Integral transforms (Laplace, Fourier) map functions between domains.

<a id='calculustransformsconvolution'></a>

---

<u>**Convolution**</u>

If $f$ and $g$ are functions from $\mathbb{R}$ to $\mathbb{R}$, their convolution $f * g$ is a new function defined by:

$$(f * g)(x) = \int_{-\infty}^{\infty} f(t)\, g(x - t)\, dt$$

The convolution extends to discrete sequences. Given sequences $a = (a_0, a_1, a_2, \ldots)$ and $b = (b_0, b_1, b_2, \ldots)$, their convolution $c = a * b$ is:

$$c_n = \sum_{k=0}^{n} a_k \, b_{n-k}$$

```python
# discrete convolution of a = [1, 2, 3] and b = [4, 5, 6]
a = [1, 2, 3]
b = [4, 5, 6]
n_out = len(a) + len(b) - 1
c = []
for n in range(n_out):
    c_n = 0
    for k in range(n + 1):
        if k < len(a) and 0 <= n - k < len(b):
            c_n += a[k] * b[n - k]
    c.append(c_n)
print(f'a = {a}')
print(f'b = {b}')
print(f'a * b = {c}')
```

<a id='calculustransformslaplace'></a>

---

<u>**Laplace Transform**</u>

Let $f : \mathbb{R}_+ \to \mathbb{R}$. The Laplace transform of $f$ is denoted $\mathcal{L}f$. It is a new function defined on the nonnegative real numbers:

$$(\mathcal{L}f)(s) = \int_0^{\infty} f(t)\, e^{-st}\, dt$$

The inverse Laplace transform is denoted $\mathcal{L}^{-1}$.

```python
# Laplace transform of f(t) = e^{-t} is 1/(s+1)
def laplace(f, s, upper=50, n=50000):
    dt = upper / n
    return sum(f((i+0.5)*dt) * math.exp(-s*(i+0.5)*dt) * dt for i in range(n))

f = lambda t: math.exp(-t)
for s in [0.5, 1.0, 2.0, 5.0]:
    Lf = laplace(f, s)
    print(f'(Lf)({s}) = {Lf:.8f}   exact: 1/(s+1) = {1/(s+1):.8f}')
```

<a id='calculustransformsfourier'></a>

---

<u>**Fourier Transform**</u>

Let $f : \mathbb{R} \to \mathbb{R}$ (or more generally, $f : \mathbb{R} \to \mathbb{C}$). The Fourier transform of $f$ is denoted $\mathcal{F}f$. It is a new function $F : \mathbb{R} \to \mathbb{C}$:

$$(\mathcal{F}f)(\omega) = F(\omega) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{\infty} f(x)\, \exp[-i\omega x]\, dx$$

Variations on this definition differ in the coefficient outside the integral and/or the sign in the exponential term. The inverse Fourier transform is denoted $\mathcal{F}^{-1}$.

```python
# Fourier transform of f(x) = exp(-x^2/2) is F(w) = exp(-w^2/2)
def fourier(f, omega, lo=-10, hi=10, n=50000):
    dx = (hi - lo) / n
    total = sum(f(lo+(i+0.5)*dx) * cmath.exp(-1j*omega*(lo+(i+0.5)*dx)) * dx for i in range(n))
    return total / math.sqrt(2 * math.pi)

f = lambda x: math.exp(-x**2 / 2)
for omega in [0.0, 0.5, 1.0, 2.0]:
    F_w = fourier(f, omega)
    print(f'F({omega}) = {F_w.real:.8f}   exact = {math.exp(-omega**2/2):.8f}')
```

<a id='calculustransformsdft'></a>

---

<u>**Discrete Fourier Transform**</u>

The discrete Fourier transform applies to a finite sequence $a = (a_0, a_1, \ldots, a_{n-1})$ and produces a new sequence $A = (A_0, A_1, \ldots, A_{n-1})$:

$$A_k = \sum_{j=0}^{n-1} a_j \, e^{-j \cdot 2\pi i \cdot k / n}$$

Some authors scale this by a factor of $1/\sqrt{n}$. Discrete Fourier transforms can be efficiently computed using the fast Fourier transform (FFT) algorithm. The discrete Fourier transform is closely related to the discrete cosine transform (DCT).

```python
# DFT of a = [1, 2, 3, 4]
a = [1, 2, 3, 4]
n = len(a)
A = []
for k in range(n):
    A_k = sum(a[j] * cmath.exp(-2j * cmath.pi * j * k / n) for j in range(n))
    A.append(A_k)
print(f'a = {a}')
print('DFT(a):')
for k in range(n):
    print(f'  A[{k}] = {A[k].real:+8.4f} {A[k].imag:+8.4f}j')
```

<a id='calculustransformsseries'></a>

---

<u>**Fourier Series**</u>

The Fourier series of a periodic function $f : [-\pi, \pi] \to \mathbb{R}$ represents $f$ as a sum of sine and cosine terms:

$$f(t) = a_0 + \sum_{k=1}^{\infty} \left( a_k \cos kt + b_k \sin kt \right)$$

This can be written in exponential form:

$$f(t) = \sum_{k=-\infty}^{\infty} \hat{f}(k)\, e^{ikt}$$

The numbers $\hat{f}(k)$ are the Fourier coefficients, computed as:

$$a_0 = \frac{1}{2\pi} \int_{-\pi}^{\pi} f(t)\, dt, \qquad a_k = \frac{1}{\pi} \int_{-\pi}^{\pi} f(t) \cos(kt)\, dt, \qquad b_k = \frac{1}{\pi} \int_{-\pi}^{\pi} f(t) \sin(kt)\, dt$$

```python
# Fourier coefficients of f(t) = t on [-pi, pi]
# a_0 = 0,  a_k = 0,  b_k = 2*(-1)^{k+1} / k
n_pts = 50000
dt = 2 * math.pi / n_pts
print('Fourier coefficients of f(t) = t:\n')
for k in range(0, 5):
    a_k = sum((-math.pi+(i+0.5)*dt) * math.cos(k*(-math.pi+(i+0.5)*dt)) for i in range(n_pts)) * dt / math.pi
    if k == 0:
        a_k /= 2
        print(f'a_0 = {a_k:+.8f}   (exact: 0)')
    else:
        b_k = sum((-math.pi+(i+0.5)*dt) * math.sin(k*(-math.pi+(i+0.5)*dt)) for i in range(n_pts)) * dt / math.pi
        exact_b = 2 * (-1)**(k+1) / k
        print(f'a_{k} = {a_k:+.8f}   b_{k} = {b_k:+.8f}   (exact b_{k} = {exact_b:+.8f})')
```
