<u>**Probability**</u>

In probability theory, one begins with a set of fundamental outcomes, often denoted $\Omega$ or $S$. An *event* is a subset of $\Omega$. The probability of an event $A$ is typically denoted $P(A)$. The notations $\Pr(A)$ and $\mathbb{P}(A)$ are also used.

<a id='probstatsprobevents'></a>

---

<u>**Events**</u>

Given a sample space $\Omega$ and two events $A, B \subseteq \Omega$, the following notations are standard:

- $P(A \cap B)$ is the probability of $A$ *and* $B$. It is also sometimes denoted $P(AB)$ or $P(A \wedge B)$.
- $P(A \cup B)$ is the probability of $A$ *or* $B$. It is also sometimes denoted $P(A \vee B)$.

For finite, equally likely sample spaces, the probability of an event is:

$$P(A) = \frac{|A|}{|\Omega|}$$

```python
omega = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {1, 2, 3}
print('P(A) =', len(A) / len(omega))
print('P(B) =', len(B) / len(omega))
```

```python
omega = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {1, 2, 3}
print('P(A and B) =', len(A & B) / len(omega))
print('P(A or B) =', len(A | B) / len(omega))
```

<a id='probstatsprobconditional'></a>

---

<u>**Conditional Probability**</u>

The conditional probability of event $A$ given event $B$ is denoted $P(A \mid B)$ and is defined as:

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}$$

This measures the probability of $A$ occurring, given that $B$ has occurred.

```python
omega = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {1, 2, 3}
P_A_given_B = len(A & B) / len(B)
P_A_given_B
```

<a id='probstatsprobrandvar'></a>

---

<u>**Random Variables**</u>

Random variables (often abbreviated rv) are generally denoted with capital letters. The notation $X(\omega)$ is the value of $X$ at the point $\omega$ of the underlying probability space on which $X$ is defined. Notation such as $P(X \ge 0)$ is a shorthand for $P(\{\omega : X(\omega) \ge 0\})$.

```python
omega = {1, 2, 3, 4, 5, 6}
X = {w: w for w in omega}
# P(X >= 4) = P({w : X(w) >= 4})
favorable = {w for w in omega if X[w] >= 4}
len(favorable) / len(omega)
```

<a id='probstatsprobindicator'></a>

---

<u>**Indicator Random Variable**</u>

Given an event $A$, the notation $\mathbf{1}_A$ is an *indicator* random variable defined by:

$$\mathbf{1}_A(\omega) = \begin{cases} 1 & \text{if } \omega \in A \\ 0 & \text{otherwise} \end{cases}$$

Some authors write $I_A$ for indicator random variables.

```python
A = {2, 4, 6}
I_A = [1 if w in A else 0 for w in range(1, 7)]
print('omega:', list(range(1, 7)))
print('I_A:  ', I_A)
```

<a id='probstatsprobexpected'></a>

---

<u>**Expected Value**</u>

For a random variable $X$, its *expected value* is denoted $E(X)$. Sometimes a different style is used, e.g., $\mathcal{E}(X)$ or $\mathbb{E}(X)$. The letter $\mu$ (standing for *mean*) is often used for the expected value. If more than one random variable is under consideration, one may subscript $\mu$ with the name of the random variable: $\mu_X$ is the expected value of $X$.

In some disciplines, the notation $\langle X \rangle$ is used to denote the expected value of $X$.

For a discrete random variable taking values $x_1, x_2, \ldots, x_n$ with equal probability:

$$E(X) = \frac{1}{n}\sum_{i=1}^{n} x_i$$

```python
X = [1, 2, 3, 4, 5, 6]
p = 1 / len(X)
E_X = sum(x * p for x in X)
E_X
```

<a id='probstatsprobcondexpect'></a>

---

<u>**Conditional Expectation**</u>

The conditional expected value of $X$ given $Y$ is denoted $E(X \mid Y)$. This is also written $\mu_{X \mid Y}$. It represents the expected value of $X$ when the value of $Y$ is known.

```python
X = [1, 2, 3, 4, 5, 6]
# E(X | X > 3): restrict to outcomes where X > 3
B = [x for x in X if x > 3]
E_X_given_B = sum(B) / len(B)
E_X_given_B
```

<a id='probstatsprobvariance'></a>

---

<u>**Variance**</u>

For a random variable $X$, the *variance* of $X$ is denoted $\text{Var}(X)$ and defined by:

$$\text{Var}(X) = E\left[(X - \mu)^2\right] = E(X^2) - E(X)^2$$

The variance is often denoted $\sigma^2$. It may be subscripted with the name of the random variable: $\sigma_X^2$.

```python
X = [1, 2, 3, 4, 5, 6]
mu = sum(X) / len(X)
Var_X = sum((x - mu)**2 for x in X) / len(X)
print('E(X) =', mu, ', Var(X) =', round(Var_X, 4))
```

<a id='probstatsprobstddev'></a>

---

<u>**Standard Deviation**</u>

The square root of the variance is known as the *standard deviation* and is denoted $\sigma$:

$$\sigma = \sqrt{\text{Var}(X)}$$

```python
X = [1, 2, 3, 4, 5, 6]
mu = sum(X) / len(X)
sigma = math.sqrt(sum((x - mu)**2 for x in X) / len(X))
sigma
```

<a id='probstatsprobcovariance'></a>

---

<u>**Covariance**</u>

Given two random variables $X$ and $Y$, their *covariance* is:

$$\text{Cov}(X, Y) = E(XY) - E(X)\,E(Y)$$

A positive covariance indicates that $X$ and $Y$ tend to increase together; a negative covariance indicates an inverse relationship.

```python
X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]
n = len(X)
E_XY = sum(X[i] * Y[i] for i in range(n)) / n
Cov_XY = E_XY - (sum(X) / n) * (sum(Y) / n)
Cov_XY
```

<a id='probstatsprobcovmatrix'></a>

---

<u>**Covariance Matrix**</u>

If $\mathbf{X}$ is a random vector,

$$\mathbf{X} = \begin{bmatrix} X_1 \\ X_2 \\ \vdots \\ X_n \end{bmatrix}$$

then the *covariance matrix* of $\mathbf{X}$ is often denoted $\Sigma$. This is an $n \times n$ matrix whose $i, j$-entry is $\text{Cov}(X_i, X_j)$.

```python
X1 = [1, 2, 3, 4, 5]
X2 = [2, 4, 5, 4, 5]
n = len(X1)
cov = lambda A, B: sum(A[i]*B[i] for i in range(n))/n - (sum(A)/n)*(sum(B)/n)
Sigma = [[round(cov(a, b), 4) for b in [X1, X2]] for a in [X1, X2]]
Sigma
```

<a id='probstatsprobcorrelation'></a>

---

<u>**Correlation Coefficient**</u>

Given two random variables $X$ and $Y$, their *correlation coefficient* is:

$$\text{Corr}(X, Y) = \frac{\text{Cov}(X, Y)}{\sigma_X \, \sigma_Y}$$

The correlation coefficient is always in the range $[-1, 1]$.

```python
X, Y = [1, 2, 3, 4, 5], [2, 4, 5, 4, 5]
n = len(X)
mu_X, mu_Y = sum(X)/n, sum(Y)/n
cov_xy = sum((X[i] - mu_X) * (Y[i] - mu_Y) for i in range(n)) / n
sig_X = math.sqrt(sum((x - mu_X)**2 for x in X) / n)
sig_Y = math.sqrt(sum((y - mu_Y)**2 for y in Y) / n)
round(cov_xy / (sig_X * sig_Y), 6)
```

<a id='probstatsprobentropy'></a>

---

<u>**Entropy**</u>

The *entropy* of a random variable $X$ is denoted $H(X)$. For discrete random variables:

$$H(X) = -\sum_{x} P(X = x) \log P(X = x)$$

where the sum is over all values $x$ that $X$ may take (and with the convention $0 \log 0 = 0$). For a continuous random variable:

$$H(X) = -\int_{-\infty}^{\infty} f(x) \log f(x) \, dx$$

where $f(x)$ is the PDF. The base of the logarithm depends on the application, with the most popular choices being $2$, $e$, and $10$.

Extensions of this notation include the *joint entropy* $H(X, Y)$ and the *conditional entropy* $H(X \mid Y)$.

```python
# entropy of a fair die (base 2)
probs = [1/6] * 6
H_X = -sum(p * math.log2(p) for p in probs if p > 0)
print(f'H(X) = {H_X:.4f} bits')
print(f'log2(6) = {math.log2(6):.4f} bits  (maximum entropy)')
```

<a id='probstatsdist'></a>

---

<u>**Distributions**</u>

Random variables are often best described by their *distribution*. Distributions, in turn, can be described in one of three principal ways, each with its own acronym.

The notation $X \sim \mathcal{D}$ indicates that the random variable $X$ has distribution $\mathcal{D}$. For example, $X \sim N(0, 1)$ states that $X$ is a standard normal random variable.

<a id='probstatsdistpmf'></a>

---

<u>**Probability Mass Function (PMF)**</u>

Discrete random variables can be specified by their *probability mass function*, or PMF, which specifies the probability associated with each possible value of the random variable:

$$f(x) = P(X = x)$$

```python
# PMF of a fair die
pmf = {x: 1/6 for x in range(1, 7)}
for x, p in pmf.items():
    print(f'P(X = {x}) = {p:.4f}')
```

<a id='probstatsdistcdf'></a>

---

<u>**Cumulative Distribution Function (CDF)**</u>

Real-valued, continuous random variables can be specified by their *cumulative distribution function*, or CDF:

$$F(x) = P(X \le x)$$

Most authors denote cumulative distribution functions with uppercase letters. The inverse function of the CDF is the *quantile function*, which is often denoted $Q$.

```python
# CDF of a fair die: F(x) = P(X <= x)
for x in range(1, 7):
    F_x = x / 6
    print(f'F({x}) = P(X <= {x}) = {F_x:.4f}')
```

<a id='probstatsdistpdf'></a>

---

<u>**Probability Density Function (PDF)**</u>

Real-valued, continuous random variables can also be specified by their *probability density function* $f(x)$. The relation to the cumulative distribution function $F(x)$ is $f(x) = F'(x)$, which gives:

$$P(X \in [a, b]) = \int_a^b f(x) \, dx$$

Most authors use lowercase letters that correspond to the uppercase letter for the random variable's CDF.

```python
# standard normal PDF: phi(x) = (1/sqrt(2*pi)) * exp(-x^2 / 2)
def phi(x):
    return math.exp(-x**2 / 2) / math.sqrt(2 * math.pi)
for x in [-2, -1, 0, 1, 2]:
    print(f'phi({x:+d}) = {phi(x):.6f}')
```

<a id='probstatsdistiid'></a>

---

<u>**Identically Distributed and IID**</u>

A collection of random variables with the same distribution are called *identically distributed*. If a collection of random variables are (mutually) independent and identically distributed, they are said to be *iid* (or IID):

$$X_1, X_2, \ldots, X_n \overset{\text{iid}}{\sim} \mathcal{D}$$

```python
random.seed(42)
# X_1, ..., X_5 iid ~ N(0, 1)
X = [round(random.gauss(0, 1), 4) for _ in range(5)]
print('IID samples from N(0, 1):', X)
```

<a id='probstatsdistbinomial'></a>

---

<u>**Binomial Distribution**</u>

Let $n$ be a nonnegative integer and $0 \le p \le 1$. A discrete random variable $X$ has the *binomial distribution* $\text{Bin}(n, p)$ or $B(n, p)$ if its probability mass function is:

$$P(X = k) = \binom{n}{k}\, p^k\, (1 - p)^{n-k}$$

where $k \in \{0, 1, \ldots, n\}$.

```python
n, p = 10, 0.3
for k in range(n + 1):
    pmf = math.comb(n, k) * p**k * (1 - p)**(n - k)
    print(f'P(X = {k:2d}) = {pmf:.4f}')
```

<a id='probstatsdistpoisson'></a>

---

<u>**Poisson Distribution**</u>

Let $\lambda$ be a positive real number. A discrete random variable $X$ has the *Poisson distribution* $\text{Pois}(\lambda)$ provided:

$$P(X = k) = \frac{\lambda^k \, e^{-\lambda}}{k!}$$

where $k$ is a nonnegative integer. The use of $\lambda$ for the parameter (and hence the mean) of the Poisson distribution is quite common.

```python
lam = 3.0
for k in range(10):
    pmf = lam**k * math.exp(-lam) / math.factorial(k)
    print(f'P(X = {k}) = {pmf:.4f}')
```

<a id='probstatsdistexponential'></a>

---

<u>**Exponential Distribution**</u>

For a positive real number $\alpha$, the *exponential distribution* is denoted $\text{Exp}(\alpha)$. The CDF for a random variable with this distribution is:

$$F(t) = 1 - \exp(-\alpha t) \quad \text{where } t \ge 0$$

```python
alpha = 2.0
for t in [0.0, 0.25, 0.5, 1.0, 2.0, 5.0]:
    F_t = 1 - math.exp(-alpha * t)
    print(f'F({t:.2f}) = {F_t:.4f}')
```

<a id='probstatsdistnormal'></a>

---

<u>**Normal / Gaussian Distribution**</u>

A *standard normal* or *Gaussian* random variable $X$ has a probability density function typically denoted $\phi(x)$:

$$\phi(x) = \frac{1}{\sqrt{2\pi}} \exp\left\{-\frac{x^2}{2}\right\}$$

The corresponding CDF is often denoted $\Phi(x)$. It is closely related to the error function $\text{erf}(x)$:

$$\Phi(x) = \frac{1}{2}\left[1 + \text{erf}\!\left(\frac{x}{\sqrt{2}}\right)\right]$$

**For more information on the error function, see the [Functions](Functions.ipynb) notebook.**

More generally, if $\mu$ is a real number and $\sigma^2$ is a positive real number, then $N(\mu, \sigma^2)$ is the distribution of a normal random variable $X$ with mean $\mu$ and variance $\sigma^2$. Its probability density function is:

$$\frac{1}{\sigma\sqrt{2\pi}} \exp\left\{-\frac{(x - \mu)^2}{2\sigma^2}\right\}$$

A standard normal random variable has $X \sim N(0, 1)$.

The multivariable normal random variable $N_k(\boldsymbol{\mu}, \Sigma)$ is a random $k$-vector with mean $\boldsymbol{\mu}$ and covariance matrix $\Sigma$.

```python
# standard normal PDF phi(x) and CDF Phi(x)
def phi(x):
    return math.exp(-x**2 / 2) / math.sqrt(2 * math.pi)
def Phi(x):
    return (1 + math.erf(x / math.sqrt(2))) / 2
for x in [-2, -1, 0, 1, 2]:
    print(f'x = {x:+d}: phi(x) = {phi(x):.6f}, Phi(x) = {Phi(x):.6f}')
```

The general normal distribution $N(\mu, \sigma^2)$ with arbitrary mean and variance:

```python
mu, sigma = 5.0, 2.0
def normal_pdf(x, mu, sigma):
    return math.exp(-(x - mu)**2 / (2 * sigma**2)) / (sigma * math.sqrt(2 * math.pi))
for x in [1, 3, 5, 7, 9]:
    print(f'f({x}) = {normal_pdf(x, mu, sigma):.6f}')
```

<a id='probstatsdistchisquared'></a>

---

<u>**Chi-Squared Distribution**</u>

The *chi-squared* distribution is denoted $\chi^2$; more specifically, the chi-squared distribution with $k$ degrees of freedom is denoted $\chi^2_k$. If $X_1, X_2, \ldots, X_k$ are IID standard normal random variables, then the sum of their squares $X = X_1^2 + \cdots + X_k^2$ is chi-squared distributed $\chi^2_k$. Its probability density function is:

$$f(x) = \frac{1}{2^{k/2}\,\Gamma(k/2)}\;x^{k/2 - 1}\;e^{-x/2}$$

where $\Gamma$ is the gamma function, satisfying $\Gamma(n) = (n-1)!$ for positive integers.

**For more information on the gamma function, see the [Functions](Functions.ipynb) notebook.**

```python
k = 5
def chi2_pdf(x, k):
    return x**(k/2 - 1) * math.exp(-x/2) / (2**(k/2) * math.gamma(k/2))
for x in [1, 3, 5, 7, 10]:
    print(f'f({x}) = {chi2_pdf(x, k):.6f}')
```

<a id='probstatsdiststudentt'></a>

---

<u>**Student's $t$-Distribution**</u>

This distribution is denoted $t_\nu$, where $\nu$ is a positive integer. It is given by $Z \,/\, \sqrt{V/\nu}$ where $Z$ and $V$ are independent random variables, $Z$ is a standard normal, and $V$ is $\chi^2_\nu$. Its probability density function is:

$$f(x) = \frac{\Gamma\!\left(\frac{\nu + 1}{2}\right)}{\sqrt{\nu\pi}\;\Gamma\!\left(\frac{\nu}{2}\right)}\left(1 + \frac{x^2}{\nu}\right)^{-(\nu + 1)/2}$$

```python
nu = 5
def t_pdf(x, nu):
    c = math.gamma((nu + 1) / 2) / (math.sqrt(nu * math.pi) * math.gamma(nu / 2))
    return c * (1 + x**2 / nu) ** (-(nu + 1) / 2)
for x in [-2, -1, 0, 1, 2]:
    print(f'f({x:+d}) = {t_pdf(x, nu):.6f}')
```

<a id='probstatsdistfdist'></a>

---

<u>**$F$-Distribution**</u>

Also known as the Fisher–Snedecor distribution, this is written $F(\nu_1, \nu_2)$ where the $\nu_i$ are positive integers. It is defined as:

$$F = \frac{X_1 / \nu_1}{X_2 / \nu_2}$$

where $X_1, X_2$ are independent $\chi^2$-random variables with $\nu_1, \nu_2$ degrees of freedom, respectively.

```python
random.seed(42)
v1, v2 = 5, 10
# simulate F-distributed values from chi-squared ratio
for _ in range(5):
    X1 = sum(random.gauss(0, 1)**2 for _ in range(v1))
    X2 = sum(random.gauss(0, 1)**2 for _ in range(v2))
    print(f'F = {(X1 / v1) / (X2 / v2):.4f}')
```

---

<u>**Convergence of Random Variables**</u>

Let $X$ and $X_1, X_2, \ldots$ be random variables. There are various notions of the sequence $X_n$ converging to $X$, each with its own notation. Some authors place the notation as a subscript to the arrow rather than above, e.g., $X_n \to_p X$.

<a id='probstatsconvdist'></a>

---

<u>**Convergence in Distribution**</u>

*Convergence in distribution* (also called *convergence in law*) is denoted in various ways:

$$X_n \xrightarrow{d} X \qquad X_n \xrightarrow{D} X \qquad X_n \xrightarrow{L} X \qquad \mathcal{L}(X_n) \to \mathcal{L}(X)$$

The Central Limit Theorem is a classic result: if $X_1, X_2, \ldots$ are IID with mean $\mu$ and variance $\sigma^2$, then:

$$\frac{\sqrt{n}\,(\bar{X}_n - \mu)}{\sigma} \xrightarrow{d} N(0, 1)$$

```python
random.seed(42)
# CLT: standardized mean of Uniform(0,1) approaches N(0,1)
for n in [10, 100, 1000]:
    zs = [(sum(random.random() for _ in range(n))/n - 0.5) * (12*n)**0.5
          for _ in range(500)]
    within = sum(1 for z in zs if abs(z) < 1.96) / 500
    print(f'n = {n:4d}: P(|Z| < 1.96) = {within:.3f}  (expect ~0.950)')
```

<a id='probstatsconvprob'></a>

---

<u>**Convergence in Probability**</u>

*Convergence in probability* is denoted:

$$X_n \xrightarrow{p} X \qquad \text{or} \qquad X_n \xrightarrow{P} X$$

This means that for every $\varepsilon > 0$, $P(|X_n - X| > \varepsilon) \to 0$ as $n \to \infty$.

```python
random.seed(42)
epsilon = 0.1
for n in [10, 50, 100, 500]:
    viols = sum(1 for _ in range(1000)
                if abs(sum(random.random() for _ in range(n))/n - 0.5) > epsilon)
    print(f'n = {n:3d}: P(|X_bar - mu| > {epsilon}) = {viols/1000:.3f}')
```

<a id='probstatsconvalmostsure'></a>

---

<u>**Almost Sure Convergence**</u>

*Almost sure convergence* is denoted:

$$X_n \xrightarrow{a.s.} X$$

Almost sure convergence implies convergence in probability. The strong law of large numbers states that $\bar{X}_n \xrightarrow{a.s.} \mu$.

```python
random.seed(42)
# strong law: running average converges to mu = 0.5
values = [random.random() for _ in range(10000)]
for n in [1, 10, 100, 1000, 10000]:
    x_bar = sum(values[:n]) / n
    print(f'n = {n:5d}: X_bar = {x_bar:.6f}  (mu = 0.500000)')
```

<a id='probstatsconvltwo'></a>

---

<u>**$L^2$ Convergence**</u>

$L^2$ *convergence* is denoted:

$$X_n \xrightarrow{L^2} X$$

This means $E\!\left[(X_n - X)^2\right] \to 0$ as $n \to \infty$. $L^2$ convergence implies convergence in probability.

```python
random.seed(42)
# E[(X_bar - mu)^2] -> 0 as n -> infinity
for n in [10, 50, 100, 500]:
    mse = sum((sum(random.random() for _ in range(n))/n - 0.5)**2
              for _ in range(1000)) / 1000
    print(f'n = {n:3d}: E[(X_bar - mu)^2] = {mse:.6f}')
```

<a id='probstatsstat'></a>

---

<u>**Statistics**</u>

Given numerical data $X_1, X_2, \ldots, X_n$, the following notations are commonly used in statistics.

<a id='probstatsstataverage'></a>

---

<u>**Average**</u>

The (sample) *average* is denoted $\bar{X}$:

$$\bar{X} = \frac{1}{n}\sum_{i=1}^{n} X_i$$

```python
X = [4, 7, 13, 2, 9]
X_bar = sum(X) / len(X)
X_bar
```

<a id='probstatsstatmedian'></a>

---

<u>**Median**</u>

The (sample) *median* is denoted $\tilde{X}$ or $\hat{X}$. It is the middle value when the data are sorted. For an even number of observations, the median is the average of the two middle values.

```python
X = [4, 7, 13, 2, 9]
X_sorted = sorted(X)
n = len(X_sorted)
median = X_sorted[n // 2] if n % 2 == 1 else (X_sorted[n//2 - 1] + X_sorted[n//2]) / 2
print('sorted:', X_sorted, ', median:', median)
```

<a id='probstatsstatstddev'></a>

---

<u>**Sample Standard Deviation**</u>

The (sample) *standard deviation* is denoted $s$:

$$s = \sqrt{\frac{\sum_{i=1}^{n}(X_i - \bar{X})^2}{n - 1}}$$

The denominator $n - 1$ (rather than $n$) gives an unbiased estimate of the population variance.

```python
X = [4, 7, 13, 2, 9]
X_bar = sum(X) / len(X)
s = math.sqrt(sum((x - X_bar)**2 for x in X) / (len(X) - 1))
s
```

---

<u>**$z$-Score**</u>

A common way to standardize data is to express it as the number of standard deviations above (or below) the mean. For this one uses the *$z$-score*:

$$Z_i = \frac{X_i - \mu}{\sigma}$$

where $\mu$ is the population mean and $\sigma$ is the population standard deviation.

```python
X = [4, 7, 13, 2, 9]
mu = sum(X) / len(X)
sigma = (sum((x - mu)**2 for x in X) / len(X)) ** 0.5
Z = [round((x - mu) / sigma, 4) for x in X]
print('z-scores:', Z)
```

<u>**Order Statistics**</u>

The notation $X_{(1)}, X_{(2)}, \ldots, X_{(n)}$ represents the same values as $X_1, \ldots, X_n$ but in nondecreasing order. That is, $X_{(1)}$ is the smallest $X_i$, $X_{(n)}$ is the largest, and in general:

$$X_{(1)} \le X_{(2)} \le \cdots \le X_{(n)}$$

```python
X = [4, 7, 13, 2, 9]
X_ordered = sorted(X)
for i, x in enumerate(X_ordered, 1):
    print(f'X_({i}) = {x}')
```

<u>**Hypothesis Testing**</u>

In hypothesis testing, the *null hypothesis* is typically denoted $H_0$ and the *alternative hypothesis* by $H_A$ or $H_1$.

Incorrectly rejecting $H_0$ when it is true is known as a *type I error*, and the probability of committing such a mistake is typically denoted $\alpha$.

Accepting $H_0$ when it is false is known as a *type II error*, and the probability of this mistake is typically denoted $\beta$.

```python
# H_0: p = 0.5 (fair coin), H_1: p != 0.5
alpha = 0.05
n, heads = 100, 62
z = (heads/n - 0.5) / (0.5 / n**0.5)
print(f'z = {z:.3f}, reject H_0 at alpha = {alpha}: {abs(z) > 1.96}')
```

<u>**Error Terms**</u>

In statistical models, error (or noise) terms are often denoted with the letter $e$ or $\varepsilon$. These can be subscripted if there are multiple error terms:

$$Y_i = \beta_0 + \beta_1 X_i + \varepsilon_i$$

```python
random.seed(42)
# Y = 2X + 3 + epsilon, where epsilon ~ N(0, 0.5)
X = [1, 2, 3, 4, 5]
epsilon = [round(random.gauss(0, 0.5), 3) for _ in X]
Y = [round(2*x + 3 + e, 3) for x, e in zip(X, epsilon)]
print('epsilon:', epsilon)
print('Y:      ', Y)
```

<u>**Estimator**</u>

When one is trying to determine some numerical quantity for a population and this is done by computing an estimate from a random sample, it is conventional to place a hat on the estimator. That is, if $\theta$ is the parameter whose value is sought, then $\hat{\theta}$ is the estimator of $\theta$.

```python
random.seed(42)
# true parameter theta = 5.0, estimate from sample mean
theta = 5.0
sample = [random.gauss(theta, 1) for _ in range(50)]
theta_hat = sum(sample) / len(sample)
print(f'theta = {theta}, theta_hat = {theta_hat:.4f}')
```
